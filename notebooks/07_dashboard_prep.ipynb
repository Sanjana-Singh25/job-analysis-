```python
import pandas as pd
import ast

df = pd.read_csv("../data/jobs_with_skills.csv")
df['extracted_skills'] = df['extracted_skills'].apply(ast.literal_eval)
df['num_skills'] = df['extracted_skills'].apply(len)

# Sirf realistic salary rakho (baaki NaN ho jayegi, row delete nahi hogi)
df['salary_clean'] = df['normalized_salary'].where(df['normalized_salary'].between(15000, 400000))
df.shape
```

```python
dash = df.drop(columns=['description', 'extracted_skills']).copy()
dash['skills_list'] = df['extracted_skills'].apply(lambda x: ', '.join(x))
dash.to_csv("../data/dashboard_jobs.csv", index=False)
print("Saved dashboard_jobs.csv ✅", dash.shape)
```

```python
tech_skills = [
    'python', 'sql', 'power bi', 'tableau', 'java', 'javascript',
    'machine learning', 'deep learning', 'aws', 'azure', 'r programming',
    'sas', 'spark', 'hadoop', 'docker', 'kubernetes', 'git', 'linux',
    'c++', 'html', 'css', 'react', 'node.js', 'jira', 'scrum'
]

skills_long = (
    df[['job_id', 'title', 'state', 'industry_name', 'experience_level', 'salary_clean', 'extracted_skills']]
    .explode('extracted_skills')
    .dropna(subset=['extracted_skills'])
    .rename(columns={'extracted_skills': 'skill'})
)
skills_long['skill_type'] = skills_long['skill'].apply(lambda s: 'Tech' if s in tech_skills else 'Non-tech')

skills_long.to_csv("../data/dashboard_skills.csv", index=False)
print("Saved dashboard_skills.csv ✅", skills_long.shape)
```
