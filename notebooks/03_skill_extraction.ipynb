```python
import pandas as pd
import re
from collections import Counter

df = pd.read_csv("../data/cleaned_jobs.csv")
df.head()
```

```python
skills_list = [
    'python', 'sql', 'excel', 'power bi', 'tableau', 'java', 'javascript',
    'machine learning', 'deep learning', 'aws', 'azure', 'communication',
    'project management', 'leadership', 'data analysis', 'r programming',
    'sas', 'spark', 'hadoop', 'docker', 'kubernetes', 'git', 'linux',
    'c++', 'html', 'css', 'react', 'node.js', 'customer service',
    'sales', 'marketing', 'nursing', 'accounting', 'bookkeeping'
]
print(len(skills_list), "skills defined")
```

```python
extra_skills = [
    # healthcare
    'patient care', 'bls', 'acls', 'cpr', 'emr', 'epic', 'phlebotomy', 'triage',
    # finance / accounting
    'quickbooks', 'gaap', 'financial reporting', 'budgeting', 'auditing',
    'accounts payable', 'accounts receivable', 'payroll',
    # sales / marketing
    'crm', 'salesforce', 'seo', 'social media', 'lead generation',
    # admin / HR / education
    'microsoft office', 'scheduling', 'recruiting', 'onboarding', 'curriculum',
    # project / tech
    'agile', 'scrum', 'jira'
]
skills_list = skills_list + extra_skills
print(len(skills_list), "skills defined")
```

```python
def extract_skills(text):
    text = str(text).lower()
    found = []
    for skill in skills_list:
        pattern = r'(?<![a-z0-9])' + re.escape(skill) + r'(?![a-z0-9])'
        if re.search(pattern, text):
            found.append(skill)
    return found
```

```python
df['extracted_skills'] = df['description'].apply(extract_skills)
df[['title', 'extracted_skills']].head(10)
```

```python
from collections import Counter

# Saari skills ek list me jod do
all_skills = df['extracted_skills'].sum()
skill_counts = Counter(all_skills)

# Top 15 skills ka DataFrame banao
top_skills = pd.DataFrame(skill_counts.most_common(15), columns=['skill', 'count'])
print(top_skills)
```

```python
# Skill-extracted data ko save karo
df.to_csv("../data/jobs_with_skills.csv", index=False)
print("Saved jobs_with_skills.csv ✅")

# Top skills ko bhi alag se save kar do — dashboard ke liye kaam aayega
top_skills.to_csv("../data/top_skills.csv", index=False)
print("Saved top_skills.csv ✅")
```
